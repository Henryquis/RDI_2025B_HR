##### 1. Limpiar los documentos de caracteres que no corresponden

```python
import re

def clean_text(doc):
    # primero reemplaza los puntos por espacios
    doc = doc.replace('.', ' ')
    # luego elimina etiquetas HTML y caracteres no permitidos,
    # conservando solo los guiones que unen letras o números
    return re.sub(r'<.*?>|(?<![A-Za-z0-9])-|-(?![A-Za-z0-9])|[^A-Za-z0-9\- ]', '', doc)

```

#### 2. Stemming

```python
from nltk.stem import porter
stemmer = porter.PorterStemmer()
```

```python
def filtrar_stopwords(texto):
    stopwords = {"the", "is", "and", "to", "of", "in", "for", "a", "an"}
    return [p.lower() for p in texto.split() if p.lower() not in stopwords]
```

```python
def get_stemming(doc):
    # obtiene el documento con las stopwords eliminadas
    palabras = filtrar_stopwords(doc)
    # aplica stemming a las palabras
    return [stemmer.stem(p) for p in palabras]
```

```python
doc = df['review'][0]        # primera review
doc_limpio = clean_text(doc) # si ya tienes tu función clean_text()

print("Stemming de la review:")
print(' '.join(get_stemming(doc_limpio)))
```

5. Verificar la diferencia
Comparar el tamaño del diccionario de términos del corpus antes y después de aplicar el preprocesamiento

```python
dicc_original = set()
# Crea el diccionario con el corpus original
for docu in df['review']:
    for palabra in docu.split():
        dicc_original.add(palabra)
```

```python
len(dicc_original)
```

```python
print("Palabras antes del preprocesamiento :", len(dicc_original))
print("Palabras después del preprocesamiento:", len(doc_limpio))

reduccion = (1 - len(doc_limpio) / len(dicc_original)) * 100
print(f"Reducción del vocabulario: {reduccion:.2f}%")
```
