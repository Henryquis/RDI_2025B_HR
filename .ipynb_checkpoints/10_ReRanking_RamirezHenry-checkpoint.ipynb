# Ejercicio 10: Re-ranking

**Objetivo:** Implementar y evaluar un pipeline de Recuperación de Información en dos etapas, y analizar el impacto del re-ranking en la calidad del ranking.

## Parte 1. Preparación del corpus

* Cargar el corpus (documentos/pasajes).
* Cargar las consultas (queries).
* Cargar qrels (relevancia).

```python
from beir import util
from beir.datasets.data_loader import GenericDataLoader
import pandas as pd
```

```python
DATASET_NAME = "scifact"
DATA_DIR = "../data/beir_datasets"
url = f"https://public.ukp.informatik.tu-darmstadt.de/thakur/BEIR/datasets/{DATASET_NAME}.zip"
util.download_and_unzip(url, DATA_DIR)
```

```python
dataset_path = DATA_DIR + "/" + DATASET_NAME
corpus, queries, qrels = GenericDataLoader(dataset_path).load(split="test")
```

```python
df_corpus = (
    pd.DataFrame.from_dict(corpus, orient="index")
      .reset_index()
      .rename(columns={"index": "doc_id"})
)

df_corpus
```

```python
df_queries = (
    pd.DataFrame.from_dict(queries, orient="index", columns=["query"])
      .reset_index()
      .rename(columns={"index": "query_id"})
)

df_queries
```

```python
rows = []
for qid, docs in qrels.items():
    for doc_id, rel in docs.items():
        rows.append({
            "query_id": qid,
            "doc_id": doc_id,
            "relevance": rel
        })

df_qrels = pd.DataFrame(rows)
df_qrels
```

```python
# Elegimos una query cualquiera que tenga varios documentos relevantes
qid = "133"

print("Query:")
print(df_queries.loc[df_queries["query_id"] == qid, "query"].values[0])

print("\nDocumentos relevantes para esta query:")
df_qrels[(df_qrels["query_id"] == qid) & (df_qrels["relevance"] > 0)]
```

## Parte 2. Retrieval inicial (baseline)

* Implementar retrieval inicial con BM25
* Obtener métricas: Recall@10 nDCG@10

```python
from rank_bm25 import BM25Okapi
from beir.retrieval.evaluation import EvaluateRetrieval
import string
from tqdm import tqdm
```

```python
# Preprocesamiento básico para BM25 (tokenización del texto)
print("Procesando y tokenizando el corpus para BM25...")
corpus_ids = list(corpus.keys())
tokenized_corpus = [
    (corpus[doc_id]['title'] + " " + corpus[doc_id]['text']).lower().split() 
    for doc_id in corpus_ids
]
```

```python
# Indexación
bm25 = BM25Okapi(tokenized_corpus)
```

```python
# Recuperar los 100 documentos más relevantes antes del re-ranking
TOP_K_CANDIDATES = 100
bm25_results = {}

print("Realizando búsqueda inicial con BM25...")
for qid, query_text in tqdm(queries.items()):
    # Tokenizar la consulta
    query_tokens = query_text.lower().split()
    
    # Calcular los scores para cada documento
    scores = bm25.get_scores(query_tokens)
    
    # Crear diccionario doc_id -> score
    doc_score_map = {corpus_ids[i]: float(scores[i]) for i in range(len(corpus_ids))}
    
    # Ordenar y tomar los top-k para optimizar memoria y preparar re-ranking
    top_docs = dict(sorted(doc_score_map.items(), key=lambda x: x[1], reverse=True)[:TOP_K_CANDIDATES])
    
    bm25_results[qid] = top_docs
```

```python
# Evaluación inicial del BM25
ndcg_scores, _map_scores, recall_scores, _ = EvaluateRetrieval.evaluate(qrels, bm25_results, [1, 10, 100])
print(f"BM25 Baseline -> nDCG@10: {ndcg_scores['NDCG@10']:.4f}, Recall@10: {recall_scores['Recall@10']:.4f}")
```

## Parte 3. Implementación del re-ranking _cross-encoder_

* Re-rankear los top-k candidatos para cada query.
* Identificar qué documentos cambian de posición en el top 10

```python
from sentence_transformers import CrossEncoder
```

```python
# Cargar el modelo Cross-Encoder
cross_encoder = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2')

ce_results = {}

TOP_N_CANDIDATES = 10  # Limitar a los primeros 20 documentos por query

print("Aplicando re-ranking con Cross-Encoder (limitado)...")
for qid, query_text in tqdm(queries.items()):
    # Tomar solo los primeros N candidatos de BM25
    candidates = list(bm25_results[qid].keys())[:TOP_N_CANDIDATES]
    
    # Crear pares [consulta, título + contenido del documento]
    query_doc_pairs = [[query_text, corpus[doc_id]['title'] + " " + corpus[doc_id]['text']] for doc_id in candidates]
    
    # Predecir los scores
    predicted_scores = cross_encoder.predict(query_doc_pairs)
    
    ce_results[qid] = {doc_id: float(score) for doc_id, score in zip(candidates, predicted_scores)}
```

```python
# Mostrar cómo cambian las posiciones para una consulta de ejemplo
example_qid = next(iter(queries))  # Seleccionamos la primera query
print(f"\nConsulta: {queries[example_qid]}")
print("Top 3 documentos según BM25:", list(bm25_results[example_qid].keys())[:3])
```

```python
# Ordenar los resultados del Cross-Encoder para obtener el top 3
top_ce_docs = sorted(ce_results[example_qid].items(), key=lambda pair: pair[1], reverse=True)
print("Top 3 documentos según Cross-Encoder:", [doc_id for doc_id, _ in top_ce_docs[:3]])
```

## Parte 4. Implementación del re-ranking _LTR_

* Re-rankear los top-k candidatos para cada query.
* Identificar qué documentos cambian de posición en el top 10

```python
from sentence_transformers import SentenceTransformer, util
import torch
```

```python
from sentence_transformers import SentenceTransformer, util

# Cargar modelo Bi-Encoder
bi_encoder = SentenceTransformer('all-MiniLM-L6-v2')

bi_encoder_results = {}

TOP_N_CANDIDATES = 5  # Limitar documentos por query
SAMPLE_QUERIES = dict(list(queries.items())[:5])  # Limitar número de queries

print("Re-rankeando con Bi-Encoder (Dense Retrieval) limitado...")

for qid, query_text in tqdm(SAMPLE_QUERIES.items()):
    # Limitar candidatos
    candidate_doc_ids = list(bm25_results[qid].keys())[:TOP_N_CANDIDATES]
    candidate_texts = [corpus[did]['title'] + " " + corpus[did]['text'] for did in candidate_doc_ids]
    
    # Embedding de la query
    query_emb = bi_encoder.encode(query_text, convert_to_tensor=True)
    
    # Embeddings de documentos candidatos
    doc_embs = bi_encoder.encode(candidate_texts, convert_to_tensor=True)
    
    # Similitud coseno
    scores = util.cos_sim(query_emb, doc_embs)[0]
    
    # Guardar resultados
    bi_encoder_results[qid] = {did: float(scores[i]) for i, did in enumerate(candidate_doc_ids)}
```

## Parte 5. Evaluación post re-ranking

Calcular métricas:
* nDCG@10
* MAP
* Recall@10

```python
print("Calculando métricas finales...")

# Evaluar BM25
ndcg_scores, map_scores, recall_scores, _ = EvaluateRetrieval.evaluate(qrels, bm25_results, [1, 10])

# Evaluación Cross-Encoder
ndcg_ce_scores, map_ce_scores, recall_ce_scores, _ = EvaluateRetrieval.evaluate(qrels, ce_results, [1, 10])

# Evaluación Bi-Encoder
ndcg_bi_scores, map_bi_scores, recall_bi_scores, _ = EvaluateRetrieval.evaluate(qrels, bi_encoder_results, [1, 10])

# Construcción de la tabla comparativa
results_table = {
    "Modelo": ["BM25 (Baseline)", "Bi-Encoder (Dense)", "Cross-Encoder (Re-ranker)"],
    "nDCG@10": [ndcg_scores['NDCG@10'], ndcg_bi_scores['NDCG@10'], ndcg_ce_scores['NDCG@10']],
    "Recall@10": [recall_scores['Recall@10'], recall_bi_scores['Recall@10'], recall_ce_scores['Recall@10']],
    "MAP@10": [map_scores['MAP@10'], map_bi_scores['MAP@10'], map_ce_scores['MAP@10']]
}
```

```python
import pandas as pd

df_results = pd.DataFrame(results_table)
df_results
```
